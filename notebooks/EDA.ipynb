# Análisis exploratorio de datos

Este notebook explora un conjunto de datos de anuncios de vehículos publicados para su venta, con el objetivo de identificar visualizaciones adecuadas para una aplicación web interactiva desarrollada con Streamlit.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('../vehicles_us.csv')

df.head()
```

## Análisis de precios anunciados

Primero, examinamos la distribución de los precios anunciados de los vehículos e identificamos posibles valores extremos.

```python
df['price'].describe(percentiles=[0.95, 0.99])
```

## Distribución de precios anunciados

El siguiente histograma muestra la distribución de los precios anunciados utilizando todos los registros del conjunto de datos.

```python
fig = px.histogram(
    df,
    x='price',
    nbins=50,
    title='Advertised Price Distribution'
)

fig.show()
```

## Distribución de precios anunciados sin valores extremos

Para mejorar la visualización, se excluyen los precios anunciados por encima del percentil 99.

```python
price_limit = df['price'].quantile(0.99)

df_filtered = df[df['price'] <= price_limit]

fig = px.histogram(
    df_filtered,
    x='price',
    nbins=50,
    title='Advertised Price Distribution Without Extreme Prices'
)

fig.show()
```

## Año del modelo vs. precio anunciado

Este gráfico de dispersión explora la relación entre el año del modelo del vehículo y su precio anunciado.

```python
scatter_df = df_filtered.dropna(subset=['model_year'])

fig = px.scatter(
    scatter_df,
    x='model_year',
    y='price',
    opacity=0.4,
    title='Model Year vs Advertised Price'
)

fig.show()
```

### Relación entre año, precio y condición

Para profundizar en la relación entre el año del modelo y el precio anunciado, se utiliza la condición del vehículo como una categoría adicional. Esto permite comparar visualmente cómo se distribuyen los precios entre vehículos en distintas condiciones.

```python
fig = px.scatter(
    df_filtered.dropna(subset=['model_year']),
    x='model_year',
    y='price',
    color='condition',
    opacity=0.5,
    title='Model Year vs Advertised Price by Condition'
)

fig.show()
```

### Interpretación

Al diferenciar los anuncios por condición, se puede observar cómo las distintas categorías se distribuyen dentro de la relación entre el año del modelo y el precio anunciado. Esta visualización se utilizó como base para el gráfico de dispersión interactivo del dashboard.

## Conclusiones

El análisis exploratorio permitió observar la distribución de los precios anunciados y la relación entre el año del modelo y el precio de los vehículos.

A partir de estas visualizaciones se desarrolló un dashboard interactivo en Streamlit que permite filtrar los anuncios por tipo de vehículo, condición y transmisión, excluir precios extremos, consultar métricas generales y explorar los datos mediante un histograma, un gráfico de dispersión y una tabla de datos filtrados.
